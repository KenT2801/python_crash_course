using the rang() function to generate a series of numbers
range(1,5) will then print the numbers from 1-5
NOTE: it does not print the number 5

```python
for value in range(1,5):
    print(value)

print("\n\tnotice how 5 were not printed, cause it only prints from 1-4, within the range 1-5")
```

Make a list of the first 10 square numbers, square of each integers from 1 through 10

```python
squares = []
for number in range(1,11,1):
    square = number * number
    squares.append(square) 

print(squares)
```

the above code can be written more concisely
we append each new value directly to the list 

```python
new_squares = []
```

```python
for number in range(10,21,1):
    new_squares.append(number**2)

print(f"the squared values from 10-20 are {new_squares}")
```

working with part of a list

slicing a list 

```python
players = ['charles', 'martina', 'michael', 'forence', 'eli']
print(players[0:3])
#this will print the indices from 0 to 3, exclusively, so 0,1,2 will be printed
print(players[1:4]) #index 1-3, which means from the second one to the forth one
print(players[:3])#index 0, the first value in the list, till the third item, with index 2
print(players[2:])#index 2, all the way to the last item, that means from the third item, till the end 
```

QUESTION: how would you print the last 2 items in the list ? without knowing length of the list ? 
ANSWER: you use negative index. 
    the last item -1
    the second last -2
    the third last -3
    geting the last 3 item 
    => players[-3":]

```python
print(players[-3:])
```

```python
print(players[0:6:2])
```

```python
print(players[::-1])
```

```python
print(players[:-1]) #gives everything except the last value 
```

Defining a tuple

```python
dimensions = (200,50)
for dimension in dimensions: 
    print(dimension)
```

NOTE:  You can't modify the tuple, but you can assign new values to the same variable name
this builds a brand-new tuple , and point the name at it, the old one is just thrown away. 

```python
print(f"The original dimensions are\n\t{dimensions}")
```

```python
dimensions=(400,100)
print(f"The updated dimensions are: \n\t{dimensions}")
```
